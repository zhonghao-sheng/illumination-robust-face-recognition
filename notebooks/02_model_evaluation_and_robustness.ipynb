# Model comparison walkthrough

The full experiment uses ten outer lighting folds and three grouped inner folds. LDA+SVM, a custom CNN, and ImageNet-pretrained EfficientNet-B0 are fitted using training data only. Hyperparameters are selected by inner-fold macro F1; outer test metrics are computed after each model is frozen.


```python
import numpy as np
import torch
from src.models import LdaSvmClassifier, IlluminationCNN, build_efficientnet_b0
from src.evaluate import evaluate_predictions

rng = np.random.default_rng(42)
X = np.concatenate([rng.normal(i, 0.2, size=(8, 24, 20)) for i in range(3)]).astype(np.float32)
y = np.repeat(np.arange(3), 8)
linear = LdaSvmClassifier(C=0.01).fit(X, y)
print(evaluate_predictions(y, linear.predict(X)))
cnn = IlluminationCNN(n_classes=3, kernel_size=5)
print('CNN output:', tuple(cnn(torch.from_numpy(X[:2, None])).shape))
efficientnet = build_efficientnet_b0(n_classes=3, pretrained=False)
print('EfficientNet output classes:', efficientnet.classifier[-1].out_features)
```
